```python
import json, random
U = [json.loads(l) for l in open("../data/processed/ita_units.jsonl")]
by_id = {u["id"]: u for u in U}
for u in random.sample(U, 5):
    print(u["id"], "|", u["marginal_note"], "|", " > ".join(u["breadcrumb"]))
    print(u["text"][:400], "\n---")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame(U)
df["text_length"] = df["text"].apply(len)
#plot histogram of the length of the text
df.hist(column="text_length", bins=50, figsize=(10,5))
```

```python
p50 = df["text_length"].quantile(0.5)
p90 = df["text_length"].quantile(0.9)
maximum = df["text_length"].max()
print(f"50th percentile: {p50}, 90th percentile: {p90}, maximum: {maximum}")
longest_10 = df.sort_values(by="text_length", ascending=False).head(10)
print("10 longest units:")
for i, row in longest_10.iterrows():
    print(row["id"], "|", row["marginal_note"], "|", " > ".join(row["breadcrumb"]))
    print(row["text"][:400], "\n---")
```

```python
shortest_10 = df.sort_values(by="text_length", ascending=True).head(10)
print("10 shortest units:")
for i, row in shortest_10.iterrows():
    print(row["id"], "|", row["marginal_note"], "|", " > ".join(row["breadcrumb"]))
    print(row["text"][:400], "\n---")
```
